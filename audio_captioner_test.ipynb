## 2. Load the Audio Captioner Model
Use the `AudioCaptioner` class from our workspace module to load the pre-trained model.

```python
from audio_captioner import AudioCaptioner

# instantiate the captioner (this may download model weights the first time)
cap = AudioCaptioner()
print('AudioCaptioner loaded:', cap)
```

## 3. Preprocess Audio Input
Load an example audio or video file. The captioner will internally extract audio if necessary.

```python
import os

# example path (should point to a downloaded video/audio file)
video_path = 'data/raw_videos/sENByrMHfyk.mp4'
print('File exists:', os.path.exists(video_path))
```

## 4. Generate Audio Descriptions
Use the captioner to describe the audio from the file.

```python
description = cap.describe_video_audio(video_path)
print('Generated audio description:')
print(description)
```
